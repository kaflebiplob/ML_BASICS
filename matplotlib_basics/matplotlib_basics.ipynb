```python
import matplotlib.pyplot as plt #pyplot is a collection of function that make matplotlib work as matlab
```

```python
x=[1,2,3,4,5]
y=[10,20,25,35,45]

plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.plot(x,y,color='blue',marker='*',linestyle='dashed',linewidth=2,markersize=20)

plt.title("Chart lets go")
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.show()
```
