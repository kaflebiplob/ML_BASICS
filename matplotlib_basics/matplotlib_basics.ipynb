```python
import matplotlib.pyplot as plt #pyplot is a collection of function that make matplotlib work as matlab
```

```python
x=[1,2,3,4,5]
y=[10,20,25,35,45]

plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.plot(x,y,color='blue',marker='*',linestyle='dashed',linewidth=2,markersize=20)

plt.title("Chart lets go")
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[10,20,25,35,45]
y2=[20,30,40,50,60]

plt.plot(x,y1,label="sales 2025")
plt.plot(x,y2,label="sales 2026")

plt.legend()
plt.show()
```

```python
# Bar Chart
x=[1,2,3,4,5]
y1=[10,20,25,35,45]
# y2=[20,30,40,50,60]

plt.bar(x,y1)
plt.show()
```

```python
# Histogram
# It  is used for distribution analysis
```

```python
import random
data = [random.randint(1,10) for _ in range(100)]

plt.hist(data,bins=5)
plt.title("histogram example")
plt.show()
```

```python
# piechart

# used to show part-to-whole relationship/data
```

```python
categories=['A','B','C','D','E']
sales=[10,20,25,35,45]
plt.pie(sales,labels=categories,autopct='%1.1f%%')
plt.title("Pie chart example")
plt.show()
```

```python
# Scatterplot
# used to find the relationship between variables
```

```python
plt.scatter(y1,y2)
plt.title('Scatter plot example')
plt.show()
```
