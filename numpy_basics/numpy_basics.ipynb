```python
# Array Reshaping and Flatening
```

```python
arr1=np.array([1,2,3,4,5,6])
print(arr1)
```

```python
reshaped = arr1.reshape((6,1))
print(reshaped)
```

```python
reshaped2 = arr1.reshape(2,3)
print(reshaped2)
```

```python
# Flatening
```

```python
print(reshaped2.flatten()) # flatening changed multi dimension to single dimension
```

```python
# Array Stacking and splitting
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
```

```python
print(np.vstack((a,b))) # vertical stacking - row wise
print(np.hstack((a,b))) # horizontal stacking - column wise
```

```python
#splitting
c= np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
print(np.hsplit(c,3))
```

```python
vsplit = np.vsplit(c,2)
for s in vsplit:
    print(s)
```
